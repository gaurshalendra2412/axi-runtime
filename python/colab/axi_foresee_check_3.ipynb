# AXI: the third forward check on the saved ARC-AGI-3 runs (arms K, A, B and C): the whole front held, the actions typed
**What this is.** The same saved moves of Tests 24, 25 and 26 as the first two forward checks, played back into fresh copies of the four public games (the replay still checks every saved flag). For every move the host predicts, from the earlier moves of the same level only, where the piece goes. The second check found the mover (97.4 %) but learned almost no walls, and its stops split in two kinds: in ls20 something was in front of the piece, in sp80 and g50t nothing was. This check holds the whole front of the piece as blocking when it did not move (released when the piece steps onto a cell), and types an action as stationary when it does nothing on open floor. It runs four ways on the same moves: both changes off (the second check), the front only, the typing only, both. Page: `docs/FORESEE_CHECK_3.md` (8 reading lines, locked before this runs). Comes from RFC-001 of Rajnish Choubey (a refused change is held state, released when the change goes through), read in game terms; for Sir to confirm. The two ideas were suggested by AI-written texts that are not Sir's.

**It calls no model, needs no key, no Google Drive and costs nothing.** It plays nothing new.

**What it does not show:** that a path-finder built on this would clear a level, or what the goal of a level is.

### Cell 2: the games (the toolkit takes an anonymous key by itself)

```python
import json, arc_agi
from axi.experiments import gemini_arc as G
from axi.experiments import ledger_replay as LR
from axi.experiments import foresee3 as F
ARC = arc_agi.Arcade()
def make_env(game): return ARC.make(game, seed=G.GAME_SEED)
RES = "/content/axi/docs/results/"
FILES = [("axi_gemini_arc_default.json", "K"), ("axi_gemini_arc_default.json", "A"), ("axi_gemini_arc_b_default.json", "B"), ("axi_gemini_arc_c_default.json", "C")]
print("ready:", [f"{f} arm {a}" for f, a in FILES])
```

### Cell 3: replay each arm and predict every move from the moves before it. If a line says `ReplayMismatch`, the saved moves did not replay into the game (copy the whole message)

```python
per_arm, per_game, ok = {}, {v: {} for v in F.VARIANTS}, {}
for f, arm in FILES:
    print("=" * 100); print("ARM", arm, "of", f); print("=" * 100)
    try:
        s = LR.run_file(RES + f, arm, make_env)
        vs = F.from_session(s); per_arm[arm] = vs; ok[arm] = True
        print(F.text(vs))
        for v, pg in vs.items():
            for g, c in pg.items(): per_game[v][g] = F.combine([per_game[v].get(g, {}), c])
    except Exception as e:
        print(type(e).__name__, ":", e); ok[arm] = False; per_arm[arm] = {"error": f"{type(e).__name__}: {e}"}
```

### Cell 4: the reading. The eight lines of the page, over the four arms together

```python
checks = F.verdict(per_game if all(ok.values()) else None, all(ok.values()) if ok else None)
print("ALL FOUR ARMS TOGETHER"); print(F.text(per_game))
print()
for i, t, x, v in checks: print(i, t, "|", {True: "yes", False: "NO", None: "not read"}[v], "|", x)
print("headline:", F.headline(checks))
```

### Cell 5: save the numbers and download the file. Send me the printout above and this file

```python
from google.colab import files
json.dump(dict(per_arm=per_arm, per_game=per_game, replays=ok, checks=checks), open("foresee_check_3_all.json", "w"), indent=1, default=str)
files.download("foresee_check_3_all.json")
```
