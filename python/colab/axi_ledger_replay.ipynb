# AXI: the transition ledger run over the saved ARC-AGI-3 runs of Test 24 and Test 25
**What this is.** The saved moves of the earlier runs are played back into fresh copies of the four public ARC-AGI-3 games, and the ledger of Update 92 watches every move. It tells what the host can know exactly from the screens: which objects moved, changed colour, vanished or appeared; the moves where the same screen and the same action gave different results; what each action does; and the host time per move.

**It calls no model, needs no key, no Google Drive and costs nothing.** It does not play anything new: the replay still checks every saved flag and refuses a doctored row. The structure it reads from is Rajnish Choubey's RFC-001 (the reading in game terms is for Sir to confirm, see `docs/HISTORY_LEDGER.md`).

**What it does not show:** that any of this makes a model clear more levels. It is the first look at the ledger on real screens; the result is not known before this runs.

### Cell 2: the games (the toolkit takes an anonymous key by itself)

```python
import json, arc_agi
from axi.experiments import gemini_arc as G
from axi.experiments import ledger_replay as LR
ARC = arc_agi.Arcade()
def make_env(game): return ARC.make(game, seed=G.GAME_SEED)
RES = "/content/axi/docs/results/"
FILES = [("axi_gemini_arc_default.json", "K"), ("axi_gemini_arc_default.json", "A"), ("axi_gemini_arc_b_default.json", "B")]
print("ready:", [f"{f} arm {a}" for f, a in FILES])
```

### Cell 3: the replay. One arm at a time; it prints one block per game. If a line says `ReplayMismatch`, the saved moves did not replay into the game (copy the whole message)

```python
out = {}
for f, arm in FILES:
    print("=" * 100); print("ARM", arm, "of", f); print("=" * 100)
    try:
        s = LR.run_file(RES + f, arm, make_env)
        print(LR.text(s)); out[arm] = LR.summarize(s)
    except Exception as e:
        print(type(e).__name__, ":", e); out[arm] = {"error": f"{type(e).__name__}: {e}"}
```

### Cell 4: save the numbers and download the file. Send me the printout above and this file

```python
from google.colab import files
json.dump(out, open("ledger_replay_all.json", "w"), indent=1, default=str)
files.download("ledger_replay_all.json")
```
