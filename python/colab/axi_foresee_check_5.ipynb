# AXI: the fifth forward check on the saved ARC-AGI-3 runs (arms K, A, B and C): repeats, the gap, filling the gap, the delete test
**What this is.** The same saved moves of Tests 24, 25 and 26 as the first four forward checks, played back into fresh copies of the four public games (the replay still checks every saved flag). The first four checks asked where the piece goes before a move. This one reads what the record says about the moves that were played and the ones it cannot answer: (1) how many moves in which the piece did not move were the same action from the same place that had not moved it before ("reset, repeat, pull again"); (2) every move not predicted, put under a reason; (3) two additions that fill part of that gap, switched on alone and together; (4) the third check's predictor with each of its five parts deleted alone (the Delete Test). Page: `docs/FORESEE_CHECK_5.md` (8 reading lines, locked before this runs). The patterns are read from the Cinderella Project pages in game terms; whose sentences they are is for Sir to confirm.

**It calls no model, needs no key, no Google Drive and costs nothing.** It plays nothing new and changes nothing on GitHub.

**What it does not show:** whether a model that plays uses fewer moves with a veto of repeats or with the gap filled (that is a paid test, not this), that a path-finder clears a level, or what the goal of a level is.

### Cell 2: the games (the toolkit takes an anonymous key by itself)

```python
import json, arc_agi
from axi.experiments import gemini_arc as G
from axi.experiments import ledger_replay as LR
from axi.experiments import foresee5 as F
ARC = arc_agi.Arcade()
def make_env(game): return ARC.make(game, seed=G.GAME_SEED)
RES = "/content/axi/docs/results/"
FILES = [("axi_gemini_arc_default.json", "K"), ("axi_gemini_arc_default.json", "A"), ("axi_gemini_arc_b_default.json", "B"), ("axi_gemini_arc_c_default.json", "C")]
print("ready:", [f"{f} arm {a}" for f, a in FILES])
```

### Cell 3: replay each arm and predict every move from the moves before it. If a line says `ReplayMismatch`, the saved moves did not replay into the game (copy the whole message)

```python
per_arm, per_game, ok = {}, {v: {} for v in F.VARIANTS}, {}
for f, arm in FILES:
    print("=" * 100); print("ARM", arm, "of", f); print("=" * 100)
    try:
        s = LR.run_file(RES + f, arm, make_env)
        vs = F.from_session(s); per_arm[arm] = vs; ok[arm] = True
        print(F.text(vs))
        for v, pg in vs.items():
            for g, c in pg.items(): per_game[v][g] = F.combine([per_game[v].get(g, {}), c])
    except Exception as e:
        print(type(e).__name__, ":", e); ok[arm] = False; per_arm[arm] = {"error": f"{type(e).__name__}: {e}"}
```

### Cell 4: the reading. The eight lines of the page, over the four arms together (line 4 compares arm C with arm K)


```python
allok = bool(ok) and all(ok.values())
checks = F.verdict(per_game if allok else None, allok if ok else None, per_arm if allok else None)
print("ALL FOUR ARMS TOGETHER"); print(F.text(per_game))
print()
for i, t, x, v in checks: print(i, t, "|", {True: "yes", False: "NO", None: "not read"}[v], "|", x)
print("headline:", F.headline(checks))
```

### Cell 5: save the numbers and download the file. Send me the printout above and this file

```python
from google.colab import files
json.dump(dict(per_arm=per_arm, per_game=per_game, replays=ok, checks=checks), open("foresee_check_5_all.json", "w"), indent=1, default=str)
files.download("foresee_check_5_all.json")
```
