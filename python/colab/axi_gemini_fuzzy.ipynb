# AXI test 23: a long conversation in plain sentences on Gemini 3.8 Flash, counted the way a company that runs the model counts
**What this is.** The answer to the objection "your state was a graph; real work is ambiguous text". One hosted model, **Gemini 3.8 Flash**, 400 turns of an ordinary conversation about a short list of items (owner, day, amount, status), written by a script from a hidden true list: new items, changes, **corrections**, **loose references** ("the last one we added", "whatever Priya is handling", "the same day as ..."), items dropped, and a paragraph of office chatter around every message that changes nothing. Every tenth turn is a **question** (about old and new facts, about a fact never given, about a reference that fits two items). Run two ways: **alone** (the whole chat kept and sent again at every turn) and **with AXI** (the exact list held by the framework and shown at every step, with a check before a reply touches it). Measured for both: **drift and hallucination**, the **distance** from the true list, **speed**, **latency** (first token and total, by prompt size), **tokens** and the **energy stand-in** (computed tokens and seconds; the API gives no joules). A third part, the **depth probe**, gives the model a PERFECT history at several depths.

The predictions and the reading rules were written before any call in `docs/GEMINI_FUZZY_TEST.md` (hash-locked).

The notebook refuses the next call once the saved answers cost the budget, and a second guard stops each run at a cap on tokens. Everything finished is saved and the same cell carries on.

### Cell 5: settings. Fill the lines at the top. PAID is True only after you turned billing on. The budget stays empty until you have read the pilot

```python
import json, time
from google.colab import userdata, files
from axi.experiments import gemini_fuzzy as G
from axi.experiments.api_llm import QuotaStop, ServiceBusy, BudgetStop

PAID = None          # True ONLY after you turned billing on for this key in Google AI Studio
PRICE_IN = None      # dollars per 1,000,000 INPUT tokens of Gemini 3.8 Flash (the AI Studio price page)
PRICE_OUT = None     # dollars per 1,000,000 OUTPUT tokens (thinking tokens are billed as output)
PRICE_CACHED = None  # dollars per 1,000,000 CACHED input tokens (type it: without it cached tokens are counted at the full price and the cost looks far too high)
BUDGET_USD = None    # the most you allow this test to cost. Leave None until you have read the pilot in Cell 6
THINKING = None      # None = the model's own default

assert PAID in (True, False), "Set PAID: True if you turned billing on, False for a free key."
try: KEY = userdata.get("GOOGLE_API_KEY")
except Exception: raise SystemExit("No key found. Do Cell 1: a Secret named GOOGLE_API_KEY with Notebook access on.")
if PAID: assert PRICE_IN and PRICE_OUT, "Type PRICE_IN and PRICE_OUT (dollars per million tokens). The notebook has no built-in price on purpose."
TAG = str(THINKING or "default"); CACHE = "axi_gemini_fuzzy_cache_%s.jsonl" % TAG
print("reply file:", store.restore(CACHE))
guard = dict(budget_usd=BUDGET_USD, price_in=PRICE_IN, price_out=PRICE_OUT, price_cached=PRICE_CACHED) if (PAID and BUDGET_USD) else dict(price_in=PRICE_IN, price_out=PRICE_OUT, price_cached=PRICE_CACHED)
llm, MODEL = G.GL.build_chat_llm(KEY, store.path(CACHE), daily_limit=None if PAID else 20, min_interval=0.3 if PAID else 4.5, on_write=lambda: store.persist(CACHE), thinking=THINKING, **guard)
META = dict(model=MODEL, thinking=THINKING)
sp = llm.spent_usd()
print("paid:", PAID, "| model:", MODEL, "| thinking:", THINKING or "model default", "| budget:", ("$%.2f" % BUDGET_USD) if BUDGET_USD else "NOT SET (pilot only)", "| spent so far with these settings:", ("$%.2f" % sp) if sp is not None else "-")
RESET = "The free quota resets at midnight Pacific time. Then open this notebook again, run Cells 2 to 5, and run this cell again."
BUSY = "This is Google's server being busy. Wait at least 30 minutes, then run THIS SAME CELL once. Do not run it again and again."
def go(fn):
    try: return fn()
    except QuotaStop as e:
        print("\nSTOPPED FOR NOW: " + str(e))
        if isinstance(e, BudgetStop): pass
        elif isinstance(e, ServiceBusy): print(BUSY)
        else: print(RESET)
        return None
def need_budget(): assert PAID and BUDGET_USD, "Read the pilot (Cell 6), then type BUDGET_USD in Cell 5, run Cell 5 again, and run this cell again."
def save(res, name):
    p = store.path(name + ".json"); json.dump(res, open(p, "w")); store.persist(name + ".json"); return p
def persisting(name, inner):
    print("progress file:", store.restore(name + ".jsonl"))
    def cb(*a):
        inner(*a)
        if a[0] % 25 == 0 or a[0] == a[1]: store.persist(name + ".jsonl")
    return cb
def prog_k(t, n, r):
    if t % 10 == 0 or t == n:
        sp = llm.spent_usd(); print(f"turn {t}/{n} | prompt {r['prompt_tokens']:,} (cached {r['cache_tokens']:,}) | {r['seconds']:.1f} s, first token {r['first_token'] if r['first_token'] is None else round(r['first_token'], 2)} s | exact {r['reply_exact']} | spent {('$%.2f' % sp) if sp is not None else '-'}")
def prog_a(t, n, r):
    if t % 20 == 0 or t == n:
        sp = llm.spent_usd(); print(f"step {t}/{n} | prompt {r['prompt_tokens']:,} | {r['seconds']:.1f} s | held list exact {r['exact']} (distance {r['distance']}) | spent {('$%.2f' % sp) if sp is not None else '-'}")
```

### Cell 6: the PILOT. 20 real turns of each arm (cents). The first 60 steps of the AXI arm are answered by the program so the list has its full size. It prints the real size of one turn, the speed and latency, and the projected cost of the whole test at your prices

```python
assert PAID, "The pilot needs a paid key (about 40 calls)."
pk = go(lambda: G.run_kept(llm, G.SEED_PILOT, 20, store.path("fz_pilot_k.jsonl"), meta=META))
pa = go(lambda: G.run_axi(llm, G.SEED_PILOT, 80, store.path("fz_pilot_a.jsonl"), meta=META, scripted=60))
if pk and pa:
    store.persist("fz_pilot_k.jsonl"); store.persist("fz_pilot_a.jsonl")
    a_live = pa["rows"][60:]
    G.print_stats("pilot K (kept chat)", pk["rows"]); G.print_stats("pilot A (with AXI)", a_live)
    rep = G.pilot_report(pk["rows"], a_live, PRICE_IN, PRICE_OUT, PRICE_CACHED)
    print()
    for k, v in rep.items(): print(f"  {k}: {v if not isinstance(v, float) else round(v, 3)}")
    print("\nRead these before the budget: k_growth_per_turn is the real size of one turn; k_last_prompt_projected must be at least 150,000 (reading rule 3); cost_k is arm K with the cache share measured now (cost_k_nocache: with no cache discount); cost_a is arm A (400 steps).")
    print("Then type BUDGET_USD in Cell 5 (a little above cost_a + cost_k + the depth probe, or lower: the run stops by itself and what it finished stays valid).")
```

### Cell 7: ARM A, with AXI. 400 steps, the exact list shown at every step with the same chatter. A line every 20 steps

```python
need_budget()
nameA = "fz_A_seed%d_%s" % (G.SEED, TAG)
resA = go(lambda: G.run_axi(llm, G.SEED, G.STEPS, store.path(nameA + ".jsonl"), meta=META, max_tokens=G.A_TOKEN_CAP, progress=persisting(nameA, prog_a)))
if resA:
    G.print_stats("ARM A (with AXI)", resA["rows"]); G.print_kinds(resA["rows"]); G.print_ages(resA["rows"]); store.persist(nameA + ".jsonl")
    print("spent so far on this model: $%.2f" % llm.spent_usd()); files.download(save(resA, nameA))
```

### Cell 8: ARM K, the kept chat, alone. 400 turns (it stops if the next prompt would pass 950,000 tokens). A line every 10 turns. The long and costly cell; if it stops (budget, busy server, tab closed), run Cells 2 to 5 and this cell again

```python
need_budget()
nameK = "fz_K_seed%d_%s" % (G.SEED, TAG)
resK = go(lambda: G.run_kept(llm, G.SEED, G.STEPS, store.path(nameK + ".jsonl"), meta=META, max_tokens=G.K_TOKEN_CAP, progress=persisting(nameK, prog_k)))
if resK:
    print("stopped because:", resK["stopped"] or "the 400 turns were done")
    G.print_stats("ARM K (kept chat)", resK["rows"]); G.print_bands(resK["rows"]); G.print_kinds(resK["rows"]); G.print_ages(resK["rows"]); store.persist(nameK + ".jsonl")
    print("spent so far on this model: $%.2f" % llm.spent_usd()); files.download(save(resK, nameK))
```

### Cell 9: the DEPTH PROBE. A perfect history (the ideal replies) at depths 50, 100, 200 and 300, then 10 real turns at each. Cheaper than arm K; it separates the length of the chat from the chat's own mistakes

```python
need_budget()
nameD = "fz_D_seed%d_%s" % (G.SEED, TAG)
resD = go(lambda: G.run_depth(llm, G.SEED, G.D_STARTS, G.D_ASKS, store.path(nameD + ".jsonl"), meta=META, max_tokens=G.D_TOKEN_CAP, progress=lambda s, t, r: print(f"depth {s}, turn {t}: prompt {r['prompt_tokens']:,} | exact {r['reply_exact']} | {r['seconds']:.1f} s") if t % 5 == 0 else None))
if resD:
    store.persist(nameD + ".jsonl")
    fm = lambda x, f='{:,.0f}': 'n/a' if x is None else f.format(x)
    for s, v in resD["summary"].items(): print(f"start {s}: prompt about {fm(v['mean_prompt'])} | exact {fm(None if v['reply_exact'] is None else 100*v['reply_exact'])}% | applicable {fm(None if v['reply_ok'] is None else 100*v['reply_ok'])}% | {fm(v['seconds_per_step'], '{:.1f}')} s | first token p50 {fm(v['first_p50'], '{:.2f}')}")
    print("spent so far on this model: $%.2f" % llm.spent_usd()); files.download(save(resD, nameD))
```

### Cell 10: the READING. Replays every saved reply (a check that nothing was changed), prints the speed and latency tables and the 24 lines with yes / NO, and saves the report. Send me the output and the downloaded files

```python
ok = {}
try: G.replay(resK["rows"], G.SEED, "K"); ok["K"] = True
except Exception as e: print("replay K:", e); ok["K"] = False
try: G.replay(resA["rows"], G.SEED, "A"); ok["A"] = True
except Exception as e: print("replay A:", e); ok["A"] = False
if resD: G.replay_depth(resD["rows"], G.SEED)
checks = G.verdict(resK, resA, resD, ok)
print("ARM K, by the size of the prompt:"); G.print_bands(resK["rows"])
print("\nARM A, by tenths of the job:")
for t in G.tenths(resA["rows"]): print(f"  steps {t['first']}-{t['last']}: held list exact {100*t['held_exact']:.1f}% | {t['seconds_per_step']:.2f} s | first token p50 {t['first_p50']} | prompt {t['prompt_per_step']:,.0f}")
G.print_verdict(checks)
print("usage:", llm.usage(), "| spent: $%.2f" % llm.spent_usd())
files.download(save(dict(K=resK, A=resA, D=resD, checks=checks, replays=ok), "gemini_fuzzy_all_%s" % TAG))
```
