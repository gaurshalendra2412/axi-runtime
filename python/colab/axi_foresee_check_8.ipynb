# AXI: the eighth forward check on the saved ARC-AGI-3 runs (arms K, A, B and C): the name prior pooled over the four arms, the retire rule with its break-even cost, the six level clears
**What this is.** The same saved moves of Tests 24, 25 and 26 as the first seven forward checks, played back into fresh copies of the four public games (the replay still checks every saved flag). After a law had been wrong by a different step, the later law predictions were right in 60.9 % against 85.9 % before. Six levels were cleared in the saved runs and nothing looked at them. This one asks: (1) with the name prior taken from the other games of ALL FOUR arms, is it present and right, and does the name now answer more first tries (`nm`, `all7`); (2) what does a retire rule (a law wrong by a different step once, or twice, is not used again in that level) cost in right moves, what does it avoid in wrong ones, and what would a wrong move have to cost for it to pay (the break-even ratio; no cost is assumed); (3) for the six clears and the levels of 20 or more moves that were not cleared, what does the screen just before the end show against the first screen. Page: `docs/FORESEE_CHECK_8.md` (12 reading lines, locked before this runs). ### Cell 2: the games (the toolkit takes an anonymous key by itself)

```python
import json, arc_agi
from axi.experiments import gemini_arc as G
from axi.experiments import ledger_replay as LR
from axi.experiments import foresee8 as F
ARC = arc_agi.Arcade()
def make_env(game): return ARC.make(game, seed=G.GAME_SEED)
RES = "/content/axi/docs/results/"
FILES = [("axi_gemini_arc_default.json", "K"), ("axi_gemini_arc_default.json", "A"), ("axi_gemini_arc_b_default.json", "B"), ("axi_gemini_arc_c_default.json", "C")]
print("ready:", [f"{f} arm {a}" for f, a in FILES])
```

### Cell 3: replay each arm (the replay also keeps the screen before each level clear). If a line says `ReplayMismatch`, the saved moves did not replay into the game (copy the whole message)

```python
sessions, ok = {}, {}
for f, arm in FILES:
    print("=" * 100); print("ARM", arm, "of", f); print("=" * 100)
    try:
        rows, games, moves = LR.load(RES + f, arm)
        sessions[arm] = F.feed8(rows, make_env, arm, games, moves); ok[arm] = True
        print("replayed:", sum(sessions[arm].moves.values()), "moves counted;", len(sessions[arm].clears), "level clears")
    except Exception as e:
        print(type(e).__name__, ":", e); ok[arm] = False
```

### Cell 4: predict every move from the moves before it, with the name prior taken per arm (as in the seventh check) and pooled over the four arms

```python
allok = bool(ok) and all(ok.values())
pooled = per_arm_prior = clears = None
if allok:
    pooled_arms = F.from_sessions(sessions, True); prior_arms = F.from_sessions(sessions, False)
    pooled, per_arm_prior, clears = F.pool(pooled_arms), F.pool(prior_arms), F.clear_study(sessions)
print("done" if allok else "an arm did not replay: nothing is read")
```

### Cell 5: the reading. The twelve lines of the page, over the four arms together

```python
checks = F.verdict(pooled, per_arm_prior, clears, allok if ok else None)
print("ALL FOUR ARMS TOGETHER"); print(F.text(pooled, per_arm_prior, clears))
print()
for i, t, x, v in checks: print(i, t, "|", {True: "yes", False: "NO", None: "not read"}[v], "|", x)
print("headline:", F.headline(checks))
```

### Cell 6: save the numbers and download the file. Send me the printout above and this file

```python
from google.colab import files
json.dump(dict(pooled=pooled, per_arm_prior=per_arm_prior, clears=clears, replays=ok, checks=checks), open("foresee_check_8_all.json", "w"), indent=1, default=str)
files.download("foresee_check_8_all.json")
```
